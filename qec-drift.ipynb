QEC-Drift tests whether an AI agent can diagnose a change in simulated quantum hardware noise and write a decoder that recovers on held-out measurements.

This notebook reproduces the published development table and inspects one recorded evaluation. The full agent harness, simulator setup, and scoring contract are available in the public repository. This notebook uses published records and requires no model calls.

[Project webpage](https://itsloganmann.github.io/qec-drift-bench/) · [Benchmark repository](https://github.com/itsloganmann/qec-drift-bench) · [Paper](https://github.com/itsloganmann/qec-drift-bench/blob/9b45d297f56e65125d59c788ba3617b9d8867b7c/paper/workshop-20261111/QEC_Drift_Workshop_Paper.pdf)


## What the benchmark measures

1. The agent gets a baseline Stim circuit and sample measurements from before and after a possible noise change.
2. It identifies the change and writes a decoder that predicts which new runs have logical errors.
3. The grader checks the diagnosis and the decoder's recovery on a separate test set. Stim and PyMatching provide simulation and reference methods.

[Scoring specification](https://github.com/itsloganmann/qec-drift-bench/blob/9b45d297f56e65125d59c788ba3617b9d8867b7c/docs/benchmark-spec-2026-11-01.md)


## Worked example

In this recorded run, the agent found a detector whose activity moved over successive blocks of shots. The submitted decoder still fell short of the recovery threshold.

[Public trace excerpt and archived score](https://github.com/itsloganmann/qec-drift-bench/blob/9b45d297f56e65125d59c788ba3617b9d8867b7c/docs/failure-analysis/moving-fault-example.json)


```python
record = {'diagnosis_correct': True, 'frozen_gap_closed': 0.10195674562306911, 'frozen_gap_drift': -1.1637717121588087, 'passed_relative': False, 'rule_label': 'plain_matching'}
diagnosis = "correct" if record["diagnosis_correct"] else "incorrect"
result = "pass" if record["passed_relative"] else "fail"
print(f"Diagnosis: {diagnosis}")
print(f"Recovery gap closed: {record['frozen_gap_closed']:.1%}")
print(f"Joint result: {result}")
assert record["diagnosis_correct"] and not record["passed_relative"]
```
